# 📊 Analyse exploratoire des données - Prix des denrées au Bénin

Ce notebook permet d'analyser le jeu de données **`prix_benin_test_propre.csv`** afin de :
- Charger et vérifier la structure des données
- Inspecter les types de variables et les valeurs manquantes
- Analyser les prix des denrées par catégorie, marché et année
- Visualiser les principales tendances

## 1. Importation des bibliothèques nécessaires

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration visuelle des graphiques
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
pd.set_option('display.max_columns', None)

print("Bibliothèques chargées avec succès !")
```

## 2. Chargement du fichier de données

```python
# Chargement du CSV
df = pd.read_csv('prix_benin_test_propre.csv', parse_dates=['date'])

# Affichage des dimensions et des 5 premières lignes
print(f"Dimensions : {df.shape[0]} lignes et {df.shape[1]} colonnes")
df.head()
```

## 3. Structure et qualité des données

```python
# Informations sur les colonnes et types de données
df.info()

# Vérification des valeurs manquantes
print("\n--- Valeurs manquantes par colonne ---")
print(df.isnull().sum())
```

```python
# Statistiques descriptives sur les prix (médiane, moyenne, min, max, etc.)
df[['price', 'usdprice', 'annee', 'mois']].describe()
```

## 4. Analyse des catégories et des produits

```python
print(f"Nombre de catégories : {df['category'].nunique()}")
print(f"Nombre de produits différents : {df['commodity'].nunique()}")
print(f"Nombre de départements : {df['admin1'].nunique()}")
print(f"Nombre de marchés suivis : {df['market'].nunique()}")

# Répartition par catégorie
df['category'].value_counts()
```

## 5. Visualisations graphiques

```python
# 1. Répartition des relevés par catégorie de denrées
plt.figure(figsize=(10, 5))
order = df['category'].value_counts().index
sns.countplot(data=df, y='category', order=order, color='steelblue')
plt.title('Nombre de relevés par catégorie de produits')
plt.xlabel('Nombre de relevés')
plt.ylabel('Catégorie')
plt.tight_layout()
plt.show()
```

```python
# 2. Évolution du prix médian annuel des denrées
prix_annuel = df.groupby('annee')['price'].median().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=prix_annuel, x='annee', y='price', marker='o', color='darkorange', linewidth=2.5)
plt.title('Évolution du prix médian des denrées au Bénin par année')
plt.xlabel('Année')
plt.ylabel('Prix médian (FCFA)')
plt.tight_layout()
plt.show()
```
